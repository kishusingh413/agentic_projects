Set `.env` with `OPENAI_API_KEY` and `WEATHER_API_KEY`, then run the code cells. Working directory should be the project root (folder containing `travel_assistant.py`).

```python
# Optional: pip install -r requirements.txt in a terminal first
from config import Settings
from travel_assistant import build_agent_executor

settings = Settings()
executor = build_agent_executor(settings, verbose=True)
```
